```python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text)
    
    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)
    
    # Case-insensitive normalization
    text = text.casefold()
    
    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
examples = [
    "Kelly Advisory, Inc",
    "LLC Moncada Léarning Center",
    "राम मार्केटिंग प्राइवेट लिमिटेड",
    "B+ Retail Inc",
    "  ABC   Corporation  "
]

for text in examples:
    print(f"{text!r}  →  {normalize_text(text)!r}")
```

```python
CLEANED_DIR = DATASET_DIR / "cleaned"

CLEANED_TRAIN_DIR = CLEANED_DIR / "train"
CLEANED_TEST_DIR = CLEANED_DIR / "test"

CLEANED_TRAIN_DIR.mkdir(parents=True, exist_ok=True)
CLEANED_TEST_DIR.mkdir(parents=True, exist_ok=True)

print("Cleaned dataset directories created:")
print(CLEANED_TRAIN_DIR)
print(CLEANED_TEST_DIR)
```

```python
def clean_source_file(input_path, output_path, chunksize=100_000):
    """
    Clean a large source TSV file in chunks and save the result as TSV.
    Original columns are preserved, with additional *_clean columns.
    """

    first_chunk = True

    for chunk in pd.read_csv(
        input_path,
        sep="\t",
        chunksize=chunksize
    ):
        # Clean business name
        chunk["business_name_clean"] = (
            chunk["business_name"]
            .apply(normalize_text)
        )

        # Clean business address
        chunk["business_address_clean"] = (
            chunk["business_address"]
            .apply(normalize_text)
        )

        # Clean country
        chunk["country_clean"] = (
            chunk["country"]
            .apply(normalize_text)
        )

        # Write chunk
        chunk.to_csv(
            output_path,
            sep="\t",
            index=False,
            mode="w" if first_chunk else "a",
            header=first_chunk
        )

        first_chunk = False

    print(f"Saved: {output_path}")
```

```python
clean_source_file(
    TRAIN_DIR / "train_source1.tsv",
    CLEANED_TRAIN_DIR / "train_source1_clean.tsv"
)

clean_source_file(
    TRAIN_DIR / "train_source2.tsv",
    CLEANED_TRAIN_DIR / "train_source2_clean.tsv"
)

clean_source_file(
    TRAIN_DIR / "train_source3.tsv",
    CLEANED_TRAIN_DIR / "train_source3_clean.tsv"
)
```

```python
ground_truth = pd.read_csv(
    TRAIN_DIR / "train_ground_truth.tsv",
    sep="\t",
    dtype=str
)

print("Shape:", ground_truth.shape)
print("\nColumns:")
print(ground_truth.columns.tolist())

display(ground_truth.head(10))
```

```python
# Treat missing matched_entity_ids as an empty match list
ground_truth["match_count"] = (
    ground_truth["matched_entity_ids"]
    .fillna("")
    .apply(lambda x: 0 if x == "" else len(x.split(",")))
)

print("Match count distribution:")
print(
    ground_truth["match_count"]
    .value_counts()
    .sort_index()
)

print("\nSummary:")
print(ground_truth["match_count"].describe())

print("\nSingletons (0 matches):")
print((ground_truth["match_count"] == 0).sum())

print("\nEntities with at least one match:")
print((ground_truth["match_count"] > 0).sum())

print("\nMaximum matches for one S1 entity:")
print(ground_truth["match_count"].max())
```

```python
def classify_matches(match_string):
    if pd.isna(match_string) or match_string == "":
        return 0, 0

    matches = match_string.split(",")

    s2_count = sum(x.startswith("S2-") for x in matches)
    s3_count = sum(x.startswith("S3-") for x in matches)

    return s2_count, s3_count


match_types = ground_truth["matched_entity_ids"].apply(
    classify_matches
)

ground_truth["s2_match_count"] = match_types.str[0]
ground_truth["s3_match_count"] = match_types.str[1]

print("Total S2 matches:", ground_truth["s2_match_count"].sum())
print("Total S3 matches:", ground_truth["s3_match_count"].sum())

print("\nS2 match count distribution:")
print(ground_truth["s2_match_count"].value_counts().sort_index())

print("\nS3 match count distribution:")
print(ground_truth["s3_match_count"].value_counts().sort_index())
```

```python
s2_only = (
    (ground_truth["s2_match_count"] > 0) &
    (ground_truth["s3_match_count"] == 0)
).sum()

s3_only = (
    (ground_truth["s2_match_count"] == 0) &
    (ground_truth["s3_match_count"] > 0)
).sum()

both = (
    (ground_truth["s2_match_count"] > 0) &
    (ground_truth["s3_match_count"] > 0)
).sum()

neither = (
    (ground_truth["s2_match_count"] == 0) &
    (ground_truth["s3_match_count"] == 0)
).sum()

print("S2 only :", s2_only)
print("S3 only :", s3_only)
print("Both    :", both)
print("Neither :", neither)

print("\nCheck total:", s2_only + s3_only + both + neither)
print("Expected   :", len(ground_truth))
```

```python
# Take 20 S1 entities that have at least one match
sample_gt = ground_truth[
    ground_truth["match_count"] > 0
].sample(20, random_state=42)

# Load only the required columns from the cleaned training files
s1_sample = pd.read_csv(
    CLEANED_TRAIN_DIR / "train_source1_clean.tsv",
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_name_clean",
        "business_address",
        "business_address_clean",
        "country"
    ]
)

# Create lookup dictionaries for the sampled S1 records
s1_lookup = s1_sample.set_index("entity_id").to_dict("index")

# Load S2 and S3 only for lookup
s2_sample = pd.read_csv(
    CLEANED_TRAIN_DIR / "train_source2_clean.tsv",
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_name_clean",
        "business_address",
        "business_address_clean",
        "country"
    ]
)

s3_sample = pd.read_csv(
    CLEANED_TRAIN_DIR / "train_source3_clean.tsv",
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_name_clean",
        "business_address",
        "business_address_clean",
        "country"
    ]
)

s2_lookup = s2_sample.set_index("entity_id").to_dict("index")
s3_lookup = s3_sample.set_index("entity_id").to_dict("index")

for _, row in sample_gt.iterrows():

    s1_id = row["source1_entity_id"]
    
    print("\n" + "=" * 80)
    print("S1:", s1_id)
    
    s1 = s1_lookup[s1_id]
    
    print("S1 Name   :", s1["business_name"])
    print("S1 Address:", s1["business_address"])
    print("S1 Country:", s1["country"])

    for match_id in row["matched_entity_ids"].split(","):
        
        if match_id.startswith("S2-"):
            match = s2_lookup.get(match_id)
            source = "S2"
        else:
            match = s3_lookup.get(match_id)
            source = "S3"

        if match:
            print(f"\n{source}:", match_id)
            print("Name   :", match["business_name"])
            print("Address:", match["business_address"])
            print("Country:", match["country"])
```
